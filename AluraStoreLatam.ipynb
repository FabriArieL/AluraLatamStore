#PROPÓSITO DEL ANÁLISIS

El objetivo de este análisis es ayudar al Sr. Juan a determinar qué tienda debería vender de sus 4 tiendas, basándonos en diversos factores clave. Primero, analizaremos la facturación, que es crucial para identificar la tienda con mayor ingreso y, por lo tanto, el mayor potencial de rentabilidad. A continuación, revisaremos la venta por categoría de productos, lo cual nos permitirá comprender qué categorías son más populares y generan más ventas, ayudando a identificar las fortalezas de cada tienda. Evaluaremos también la calificación promedio de la tienda, dado que una tienda con altas calificaciones probablemente tendrá más lealtad de los clientes y un mayor potencial de crecimiento a largo plazo. Es fundamental analizar los productos más y menos vendidos, ya que esto revela el desempeño de los productos específicos, lo que nos ayudará a identificar aquellos que pueden no estar funcionando y los que tienen más potencial. Finalmente, revisaremos los envíos promedio, ya que evaluar el costo de envío es clave para entender la rentabilidad de cada tienda y si los gastos en logística están afectando demasiado las ganancias.

Al considerar todos estos aspectos, se podrá tomar una decisión estratégica sobre cuál tienda tiene el mejor desempeño global, maximizando tanto la rentabilidad como la satisfacción del cliente.

### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
import matplotlib.pyplot as plt

# Sumar los precios
total1 = tienda['Precio'].sum()
total2 = tienda2['Precio'].sum()
total3 = tienda3['Precio'].sum()
total4 = tienda4['Precio'].sum()

# Crear listas para graficar
nombres_tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
totales = [total1, total2, total3, total4]

# Graficar
plt.figure(figsize=(8, 5))
barras = plt.bar(nombres_tiendas, totales, color='skyblue')

# Agregar los valores encima de cada barra
for barra in barras:
    yval = barra.get_height()
    plt.text(barra.get_x() + barra.get_width()/2, yval + yval*0.01, f"${yval:,.2f}",
             ha='center', va='bottom', fontsize=10)

plt.title('Ventas Tiendas')
plt.xlabel('Tienda')
plt.ylabel('Ingreso Total')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

# 3. Calificación promedio de la tienda


```python
import matplotlib.pyplot as plt

# Calcular promedio y cantidad de calificaciones por tienda
promedios = [
    tienda['Calificación'].mean(),
    tienda2['Calificación'].mean(),
    tienda3['Calificación'].mean(),
    tienda4['Calificación'].mean()
]

cantidades = [
    tienda['Calificación'].count(),
    tienda2['Calificación'].count(),
    tienda3['Calificación'].count(),
    tienda4['Calificación'].count()
]

nombres_tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']

# Graficar
plt.figure(figsize=(8, 5))
barras = plt.bar(nombres_tiendas, promedios, color='mediumseagreen')

# Etiquetas: promedio y cantidad
for i, barra in enumerate(barras):
    yval = barra.get_height()
    plt.text(barra.get_x() + barra.get_width()/2, yval + 0.02,
             f"{promedios[i]:.2f}\n({cantidades[i]} calif.)",
             ha='center', va='bottom', fontsize=10)

# Personalizar gráfico
plt.ylim(0, 5)  # Si las calificaciones son de 1 a 5
plt.title('Satisfacción del cliente por tienda')
plt.xlabel('Tienda')
plt.ylabel('Calificación promedio')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

# 4. Productos más y menos vendidos

```python
import matplotlib.pyplot as plt

# Función para obtener top y bottom por conteo de productos
def obtener_top_bottom_por_conteo(df):
    resumen = df['Producto'].value_counts()
    top = resumen.head(5)
    bottom = resumen.tail(5)
    return top, bottom

# Lista de tiendas y nombres
tiendas = [tienda, tienda2, tienda3, tienda4]
nombres = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']

# Crear figura con subplots
fig, axs = plt.subplots(4, 2, figsize=(14, 16))  # 4 tiendas x 2 gráficos

for i, (df, nombre) in enumerate(zip(tiendas, nombres)):
    top, bottom = obtener_top_bottom_por_conteo(df)

    # Más vendidos
    axs[i, 0].barh(top.index[::-1], top.values[::-1], color='mediumseagreen')
    axs[i, 0].set_title(f'{nombre} - Top 5 más vendidos')
    axs[i, 0].set_xlabel('Cantidad vendida')

    # Etiquetas en las barras (top)
    for j, valor in enumerate(top.values[::-1]):
        axs[i, 0].text(valor + 0.2, j, str(valor), va='center', fontsize=9)

    # Menos vendidos
    axs[i, 1].barh(bottom.index, bottom.values, color='indianred')
    axs[i, 1].set_title(f'{nombre} - Top 5 menos vendidos')
    axs[i, 1].set_xlabel('Cantidad vendida')

    # Etiquetas en las barras (bottom)
    for j, valor in enumerate(bottom.values):
        axs[i, 1].text(valor + 0.2, j, str(valor), va='center', fontsize=9)

plt.tight_layout()
plt.show()
```
